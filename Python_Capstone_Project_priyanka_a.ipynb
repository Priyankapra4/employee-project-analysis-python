```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
employee = pd.DataFrame({
    "ID":["A001","A002","A003","A004","A005"],
    "Name":["John Alter","Alice Luxumberg","Tom Sabestine","Nina Adgra","Amy Johny"],
    "Gender":["M","F","M","F","F"],
    "City":["Paris","London","Berlin","Newyork","Madrid"],
    "Age":[25,27,29,31,30]
})
seniority = pd.DataFrame({
    "ID":["A001","A002","A003","A004","A005"],
    "Designation Level":[2,2,3,2,3]
})
project = pd.DataFrame({
    "ID":["A001","A002","A003","A004","A005","A002","A005","A003","A001","A003","A001","A004","A004","A005"],
    "Project":["Project 1","Project 2","Project 3","Project 4","Project 5","Project 6","Project 7","Project 8","Project 9","Project 10","Project 11","Project 12","Project 13","Project 14"],
    "Cost":[1002000,2000000,4500000,5500000,np.nan,680000,400000,350000,np.nan,300000,2000000,1000000,3000000,200000],
    "Status":["Finished","Ongoing","Finished","Ongoing","Finished","Failed","Finished","Failed","Ongoing","Finished","Failed","Ongoing","Finished","Finished"]
})
```

```python
employee.to_csv("Employee.csv",index=False)
seniority.to_csv("Seniority.csv",index=False)
project.to_csv("Project.csv",index=False)

employee = pd.read_csv("Employee.csv")
seniority = pd.read_csv("Seniority.csv")
project = pd.read_csv("Project.csv")

print(" Employee DataFrame ")
print(employee)
print("\n Seniority DataFrame")
print(seniority)
print("\n Project DataFrame")
print(project)
```

```python
running_total = 0
count = 0
for i in range(len(project)):
    if pd.notnull(project.loc[i,"Cost"]):
        running_total = running_total + project.loc[i,"Cost"]
        count = count + 1
    else:
        avg = running_total / count
        project.loc[i,"Cost"] = avg
print("Task 2 Output:")
print(project)
```

```python
running_total = 0
count = 0

for i in range(len(project)):
    if pd.notnull(project.loc[i, "Cost"]):
        running_total = running_total + project.loc[i, "Cost"]
        count = count + 1
    else:
        if count > 0:
            avg = running_total / count
            project.loc[i, "Cost"] = avg
            running_total = running_total + avg
            count = count + 1
        else:
           project.loc[i, "Cost"] = 0

print("Task 2 Output:")
print(project)
```

```python
running_total = 0
count = 0
for i in range(len(project)):
    cost = project.loc[i, "Cost"]
    if pd.notnull(cost):
        running_total += cost
        count += 1
    else:
        avg = running_total / count
        project.loc[i, "Cost"] = avg
        running_total += avg
        count += 1

print("Task 2 Output")
print(project)
```

```python
employee[["First Name", "Last Name"]] = employee["Name"].str.split(" ", n=1, expand=True)
employee = employee.drop(columns=["Name"])
print(employee.head())
```

```python
if "Name" in employee.columns:
    employee[["First Name", "Last Name"]] = employee["Name"].str.split(" ", n=1, expand=True)
    employee = employee.drop(columns=["Name"])
print("Task 3 Output")
print(employee.head())
```

```python
final = pd.merge(employee, project,on="ID", how ="inner")
final = pd.merge(final,seniority,on="ID", how ="inner")
print("Task 4 Output: Final DataFrame")
print(final.head())
```

```python
merged_df = pd.merge(employee, project, on="ID", how="inner")

# 2. Merge the third dataframe into the combined dataframe
final = pd.merge(merged_df, seniority, on="ID", how="inner")

# 3. Verify the final joined dataframe
print("Combined Dataframe Output:")
print(final.head())
```

```python
final["Bonus"] = np.where(final["Status"]=="Finished",final["Cost"]*0.05,0)
print("Task 5 Output: Bonus")
print(final)
```

```python
final["Bonus"] = np.where(final["Status"]=="Finished",final["Cost"]*0.05,0)
print("Task 5 Output: Bonus of the Finished status employees")
print(final)
```

```python
failed_emp = final["Status"].str.lower() == "failed"
final.loc[failed_emp, "Designation Level"] = final.loc[failed_emp, "Designation Level"] + 1
final = final[final["Designation Level"] <= 4]
print("Task 6 Output: ")
print(final)
```

```python
gender_titles = {"M": "Mr. ", "F": "Mrs. "}
final["First Name"] = final["Gender"].map(gender_titles) + final["First Name"]
final = final.drop(columns=["Gender"])
print("Task 7 Output:")
print(final)
```

```python
if "Gender" in final.columns:
  final["First Name"] = np.where(final["Gender"]=="M","Mr. "+final["First Name"],"Mrs. "+final["First Name"])
  final.drop("Gender",axis=1,inplace=True)
print("Task 7 Output:")
print(final)
```

```python
if "Gender" in final.columns:
    gender_titles = {"M": "Mr. ", "F": "Mrs. "}
    final["First Name"] = final["Gender"].map(gender_titles) + final["First Name"]
    final = final.drop(columns=["Gender"])

print("Task 7 Output:")
print(final)
```

```python
for i in final.index:
  if final.loc[i, "Age"] > 29:
    final.loc[i, "Designation Level"] = final.loc[i, "Designation Level"] - 1

print("Task8 Output:")
print(final)
```

```python
TotalProjCost = final.groupby(["ID","First Name"])["Cost"].sum().reset_index()
TotalProjCost.rename(columns={"Cost":"Total Cost"},inplace=True)
print("Task 9 Output \n")
print(TotalProjCost)
```
